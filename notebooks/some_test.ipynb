```julia
using Pkg
Pkg.activate("/scratch-1/cogier/hydraulic_barriers/")
using Rasters
using ArchGDAL
using WhereTheWaterFlowsSubglacially, WhereTheWaterFlows
const WWFS = WhereTheWaterFlowsSubglacially
const WWF = WhereTheWaterFlows
using Plots
```

```julia
datadir_in = "/scratch-3/cogier/data/DEM_input"
datadir_out = "/scratch-3/cogier/data/DEM_output"
```

```julia
bed_lv95 = Raster(joinpath(datadir_in, "grid_bedrock_10m.tif"))
bed_lambert = Raster(joinpath(datadir_in, "Bedrock.tif"))
ice_thickness_2024_lv95 = Raster(joinpath(datadir_in,"grid_ice_thickness_10m.tif"));
ice_thickness_2024_lambert = Raster(joinpath(datadir_in, "IceThickness.tif"));
```

### WWFS in lv95 dem for 2024 ###

```julia
#surface 2024

dx= step(dims(bed_lv95)[1])


(area_2024, slen_2024, dir, nout, nin, pits, c, bnds),
(sc_locs, kappas, diro, phi_2024),
(lakes, lakes_free_surf_2024_lv95),
sinkout = WWFS.waterflows_subglacial(bed_lv95 .+ ice_thickness_2024_lv95, bed_lv95, dx, gamma=[0,WWFS.GAMMA][1], bnd_as_sink=true, drain_pits=true);

sum(lakes_free_surf_2024_lv95[.!isnan.(lakes_free_surf_2024_lv95)])*dx*dx
```

### WWFS in lambert (lv95 reprojected) dem for 2024 ###

```julia
#surface 2024

dx= step(dims(bed_lambert)[1])


(area_2024, slen_2024, dir, nout, nin, pits, c, bnds),
(sc_locs, kappas, diro, phi_2024),
(lakes, lakes_free_surf_2024_lamb),
sinkout = WWFS.waterflows_subglacial(bed_lambert .+ ice_thickness_2024_lambert, bed_lambert, dx, gamma=[0,WWFS.GAMMA][1], bnd_as_sink=true, nan_as_sink=true, drain_pits=true);

sum(lakes_free_surf_2024_lamb[.!isnan.(lakes_free_surf_2024_lamb)])*dx*dx
```
